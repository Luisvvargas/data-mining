# Otras técnicas de sumarización

En este notebook se aplicarán tres técnicas de sumarización sobre el dataset del Titanic: asimetría, curtosis y tablas cruzadas.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)

df.head()
```

## 1. Asimetría (Skewness)

La asimetría nos permite ver si los datos están distribuidos de forma equilibrada o si se concentran más hacia alguno de los lados. En este caso se analizará la columna `Age`.

```python
age_skewness = df["Age"].skew()

age_skewness
```

El resultado de la asimetría de `Age` es positivo, aunque no es demasiado alto. Esto significa que la distribución tiene una ligera cola hacia las edades mayores.

En términos del Titanic, había una mayor concentración de pasajeros jóvenes y adultos, mientras que había menos pasajeros de edades muy altas. Es decir, los pasajeros mayores forman la cola hacia la derecha de la distribución.

## 2. Curtosis (Kurtosis)

La curtosis permite analizar qué tan pesadas son las colas de una distribución. Se aplicará sobre la columna `Fare` para observar si existen valores extremos en las tarifas pagadas por los pasajeros.

```python
fare_kurtosis = df["Fare"].kurt()

fare_kurtosis
```

La curtosis de `Fare` es bastante alta y positiva. Esto indica que la distribución tiene colas pesadas, es decir, existen algunos valores que se encuentran muy alejados de la mayoría.

Esto confirma los outliers encontrados anteriormente en la columna `Fare`, ya que la mayoría de los pasajeros pagó tarifas relativamente normales, pero algunos pagaron cantidades mucho más altas que el resto.

## 3. Tablas Cruzadas (Cross-Tabulation)

Una tabla cruzada permite comparar rápidamente dos variables categóricas. En este caso se cruzará la clase del pasajero (`Pclass`) con la variable que indica si sobrevivió (`Survived`).

```python
tabla_cruzada = pd.crosstab(df["Pclass"], df["Survived"])

tabla_cruzada
```

Al cruzar `Pclass` con `Survived` se puede observar rápidamente que los pasajeros de primera clase tuvieron una mayor cantidad de sobrevivientes, mientras que en tercera clase hubo una cantidad mucho mayor de personas que no sobrevivieron.

Esto muestra que la clase del pasajero tuvo una relación clara con la supervivencia. Los pasajeros de primera clase tuvieron mejores posibilidades de sobrevivir que los pasajeros de tercera clase.
